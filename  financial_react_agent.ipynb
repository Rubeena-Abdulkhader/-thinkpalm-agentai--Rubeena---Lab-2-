```python
import re
import gradio as gr

# =====================================================
# TOOL : Compound Interest Calculator
# =====================================================

def compound_interest(principal, rate, years):
    return principal * ((1 + rate / 100) ** years)

# =====================================================
# AGENT
# =====================================================

def financial_agent(message, history):

    try:

        numbers = re.findall(r'\d+\.?\d*', message)

        if len(numbers) < 3:
            return """
🤖 Financial Advisor Agent

Please ask a question like:

If I invest 50000 at 12% for 10 years
"""

        principal = float(numbers[0])
        rate = float(numbers[1])
        years = float(numbers[2])

        # ==================================
        # ReAct Reasoning (Internal)
        # ==================================

        thought = "User wants future investment value."

        action = (
            f"Calling Compound Interest Calculator Tool "
            f"(P={principal}, R={rate}, T={years})"
        )

        amount = compound_interest(
            principal,
            rate,
            years
        )

        observation = (
            f"Calculator returned ₹{amount:,.2f}"
        )

        # ==================================
        # Display ReAct Steps in Colab Logs
        # ==================================

        print("\n==============================")
        print("THOUGHT")
        print("==============================")
        print(thought)

        print("\n==============================")
        print("ACTION")
        print("==============================")
        print(action)

        print("\n==============================")
        print("OBSERVATION")
        print("==============================")
        print(observation)

        print("==============================\n")

        # ==================================
        # USER RESPONSE
        # ==================================

        response = f"""
🤖 Financial Advisor Agent

Investment Summary

• Amount Invested: ₹{principal:,.0f}
• Interest Rate: {rate}%
• Duration: {years:.0f} years

Projected Future Value

💰 ₹{amount:,.2f}
"""

        return response

    except Exception as e:
        return f"Error: {str(e)}"


# =====================================================
# COMPACT UI
# =====================================================

with gr.Blocks(theme=gr.themes.Soft()) as demo:

    gr.Markdown(
        """
## 🤖 Financial Advisor Agent

Ask investment-related questions.
"""
    )

    with gr.Row():
        with gr.Column(scale=1):

            chatbot = gr.Chatbot(
                height=350,
                show_label=False
            )

            msg = gr.Textbox(
                label="Your Question",
                placeholder="Example: If I invest 100000 at 12% for 12 years"
            )

            clear = gr.Button("🗑️ Clear Chat")

            def respond(message, chat_history):

                response = financial_agent(
                    message,
                    chat_history
                )

                chat_history.append(
                    (message, response)
                )

                return "", chat_history

            msg.submit(
                respond,
                [msg, chatbot],
                [msg, chatbot]
            )

            clear.click(
                lambda: [],
                None,
                chatbot
            )

demo.launch(
    share=True,
    debug=False
)
```
